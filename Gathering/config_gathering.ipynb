```python
%sql
CREATE CATALOG IF NOT EXISTS test COMMENT 'This is gathering config and CDM catalog'
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS test.gathering COMMENT 'This is the gathering schema'
```

```python
%sql
CREATE TABLE IF NOT EXISTS test.gathering.gathering_config (
  id INT PRIMARY KEY,
  channel STRING NOT NULL,
  object_name STRING NOT NULL,
  full_history STRING,
  partial_history STRING,
  accounts_config STRING,
  enable BOOLEAN NOT NULL
)
COMMENT 'This is the gathering config table'
```
