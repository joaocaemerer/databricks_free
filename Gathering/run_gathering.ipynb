```python
from pyspark.sql.functions import explode

df_gathering = spark.table('test.gathering.gathering_config')
df_gathering_enabled = df_gathering.filter(df_gathering.enable == True)

sql_result = []
for row in df_gathering_enabled.collect():
    sql_result.append({'id': row['id'], 'accounts_config_result': [ ac_row.asDict() for ac_row in spark.sql(row['accounts_config']).collect()]})

df_sql_result = spark.createDataFrame(sql_result)
df_sql_result.display()

df_gathering_enabled.join(
    df_sql_result, df_gathering_enabled.id == df_sql_result.id
).withColumn('accounts_config_result', explode('accounts_config_result')).display()
```
